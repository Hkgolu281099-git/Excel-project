```python
from google.colab import auth
import gspread
from google.auth import default
import pandas as pd
from gspread_dataframe import get_as_dataframe

# 1. Authenticate your Google account
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

# 2. Define your Google Sheet URL
sheet_url = "https://docs.google.com/spreadsheets/d/1jh8UeSiUokzElxnSYRbD1JAgN-Wqkk4Sqp9axEvNDbw/edit#gid=1728593922"

# 3. Open the spreadsheet and select the specific sheet by its gid
sh = gc.open_by_url(sheet_url)
worksheet = sh.get_worksheet_by_id(1728593922)  # Fetches the sheet with gid=1728593922

# 4. Read the data directly into a pandas DataFrame called df
df = get_as_dataframe(worksheet).dropna(how="all")

# 5. Display the first few rows
df.head()
```

```python
# 1 & 2. Number of rows and columns
print("--- 1 & 2. ROWS AND COLUMNS ---")
print(f"Number of rows: {len(df)}")
print(f"Number of columns: {len(df.columns)}\n")

# 3. Column names
print("--- 3. COLUMN NAMES ---")
print(list(df.columns))
print("\n")

# 4. Data types
print("--- 4. DATA TYPES ---")
print(df.dtypes)
print("\n")

# 5. Summary statistics
print("--- 5. SUMMARY STATISTICS ---")
# describe() gives you math stats (mean, min, max, etc.) for numerical columns
display(df.describe())
print("\n")

# 6. Unique values in categorical columns
print("--- 6. UNIQUE VALUES IN CATEGORICAL COLUMNS ---")
# This automatically finds columns that contain text or categories
categorical_columns = df.select_dtypes(include=['object', 'category']).columns

for col in categorical_columns:
    print(f"\nColumn: '{col}'")
    print(f"Total unique values: {df[col].nunique()}")
    # We print just the first 10 unique values so it doesn't overwhelm your screen
    print(f"Sample values: {df[col].dropna().unique()[:10]}")
print("\n")

# 7. First 5 rows
print("--- 7. FIRST 5 ROWS ---")
display(df.head())
```

```python
import pandas as pd

# 1. Clean Revenue & Profit columns (convert to string first, strip non-numeric symbols, convert to numeric)
for col in ['Revenue', 'Profit']:
    # Keep only digits, negative signs, and decimals
    df[col] = df[col].astype(str).str.replace(r'[^0-9.-]', '', regex=True)

    # Convert to numeric (turns empty or invalid strings into NaN)
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 2. Fill missing values (NaN) with 0 before converting to integer (or keep float if preferred)
df['Revenue'] = df['Revenue'].fillna(0).astype('int64')

# Profit contains missing values/floats; keeping as int64 or float64 depending on preference
df['Profit'] = df['Profit'].fillna(0).astype('int64')

# 3. Verify updated data types
print("--- UPDATED DATA TYPES ---")
print(df[['Revenue', 'Profit']].dtypes)
print("\n")

# 4. Display updated summary statistics for numeric columns
print("--- UPDATED SUMMARY STATISTICS ---")
display(df.describe())

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Convert Order_Date to proper datetime format
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

# 2. Extract Year-Month to group data chronologically
df['Year_Month'] = df['Order_Date'].dt.to_period('M')

# 3. Calculate total Revenue for each month
monthly_revenue = df.groupby('Year_Month')['Revenue'].sum().reset_index()

# 4. Convert Year_Month back to string format for clean plotting on x-axis
monthly_revenue['Year_Month_Str'] = monthly_revenue['Year_Month'].astype(str)

# Identify highest and lowest revenue months for quick reference
highest_month = monthly_revenue.loc[monthly_revenue['Revenue'].idxmax()]
lowest_month = monthly_revenue.loc[monthly_revenue['Revenue'].idxmin()]

print(f"Highest Revenue Month: {highest_month['Year_Month_Str']} ({highest_month['Revenue']:,})")
print(f"Lowest Revenue Month:  {lowest_month['Year_Month_Str']} ({lowest_month['Revenue']:,})\n")

# 5, 6, 7 & 8. Create and display the line chart
plt.figure(figsize=(10, 5))
plt.plot(monthly_revenue['Year_Month_Str'], monthly_revenue['Revenue'], marker='o', linewidth=2, color='b')

plt.title('Monthly Total Revenue', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Revenue', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# Group by month and calculate sum
monthly_rev = df.groupby(df['Order_Date'].dt.to_period('M'))['Revenue'].sum()

# Identify highest and lowest months
max_month = monthly_rev.idxmax()
min_month = monthly_rev.idxmin()

# Display results
print(f"Highest Revenue Month: {max_month} with Revenue of {monthly_rev[max_month]:,}")
print(f"Lowest Revenue Month:  {min_month} with Revenue of {monthly_rev[min_month]:,}")
```

```python
import matplotlib.pyplot as plt

# 1, 2, & 3. Group by Category, calculate total Revenue, and sort from highest to lowest
category_revenue = df.groupby('Category')['Revenue'].sum().reset_index()
category_revenue = category_revenue.sort_values(by='Revenue', ascending=False)

# Identify top category
top_category = category_revenue.iloc[0]
print(f"Top Category: {top_category['Category']} with Total Revenue of {top_category['Revenue']:,}\n")

# 4, 5, & 6. Create bar chart with distinct colors for each category bar
plt.figure(figsize=(10, 6))

# Generate a unique color for each category bar using Matplotlib's built-in 'tab10' palette
colors = plt.cm.tab10(range(len(category_revenue)))

plt.bar(
    category_revenue['Category'],
    category_revenue['Revenue'],
    color=colors,
    edgecolor='black'
)

plt.title('Total Revenue by Product Category', fontsize=14)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Total Revenue', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Deduplicate dataset so each Customer_ID is only counted once
unique_customers = df.drop_duplicates(subset=['Customer_ID'])

# 2. Count unique customers and calculate percentages per Customer_Type
segment_counts = unique_customers['Customer_Type'].value_counts()
segment_percentages = (unique_customers['Customer_Type'].value_counts(normalize=True) * 100).round(2)

# Combine into a clean breakdown table
segment_summary = pd.DataFrame({
    'Unique_Customers': segment_counts,
    'Percentage': segment_percentages
})

print("--- CUSTOMER TYPE BREAKDOWN ---")
print(segment_summary)
print("\n")

# 3. Create a bar chart showing unique customers by segment/type
plt.figure(figsize=(8, 5))
bars = plt.bar(
    segment_summary.index.astype(str),
    segment_summary['Unique_Customers'],
    color=plt.cm.Set2.colors[:len(segment_summary)],
    edgecolor='black'
)

# 4. Add percentage labels on top of each bar
for bar, pct in zip(bars, segment_summary['Percentage']):
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2.0,
        yval + (yval * 0.01),
        f'{pct}%',
        ha='center',
        va='bottom',
        fontsize=10,
        fontweight='bold'
    )

plt.title('Customer Distribution by Customer Type', fontsize=14)
plt.xlabel('Customer Type', fontsize=12)
plt.ylabel('Unique Customer Count', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Deduplicate dataset so each Customer_ID is counted only once
unique_customers = df.drop_duplicates(subset=['Customer_ID'])

# 2. Calculate unique customer counts per Customer_Type
segment_counts = unique_customers['Customer_Type'].value_counts()

# 3. Create a pie chart showing customer distribution
plt.figure(figsize=(7, 7))

plt.pie(
    segment_counts,
    labels=segment_counts.index,
    autopct='%1.1f%%',          # Formats percentage to 1 decimal place
    startangle=140,
    colors=plt.cm.Set2.colors[:len(segment_counts)],
    wedgeprops={'edgecolor': 'black'}
)

plt.title('Customer Distribution by Customer Type', fontsize=14)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Filter out any "unknown region" entries (case-insensitive)
df_filtered = df[df['Region'].str.strip().str.lower() != 'unknown region'].copy()

# 2. Group by Region, sum Revenue, and sort from highest to lowest
region_revenue = (
    df_filtered.groupby('Region')['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=True)  # Ascending order so the highest appears at the top of the horizontal bar chart
)

# Identify highest and lowest revenue regions
highest_region = region_revenue.iloc[-1]
lowest_region = region_revenue.iloc[0]

print(f"Highest Revenue Region: {highest_region['Region']} ({highest_region['Revenue']:,})")
print(f"Lowest Revenue Region:  {lowest_region['Region']} ({lowest_region['Revenue']:,})\n")

# 3, 4, 5, 6 & 7. Create horizontal bar chart with distinct colors
plt.figure(figsize=(10, 6))

colors = plt.cm.Set3(range(len(region_revenue)))

plt.barh(
    region_revenue['Region'],
    region_revenue['Revenue'],
    color=colors,
    edgecolor='black'
)

plt.title('Total Revenue by Region (Excluding Unknown)', fontsize=14)
plt.xlabel('Revenue', fontsize=12)
plt.ylabel('Region', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1, 2, 3 & 4. Group by Product, sum Profit, sort from highest to lowest, and select the Top 10
top_10_products = (
    df.groupby('Product')['Profit']
    .sum()
    .reset_index()
    .sort_values(by='Profit', ascending=False)
    .head(10)
)

# Reverse the order for plotting so the #1 product appears at the top of the horizontal bar chart
plot_data = top_10_products.sort_values(by='Profit', ascending=True)

# Print the top-performing product
top_product = top_10_products.iloc[0]
print(f"Highest Profit Product: {top_product['Product']} with Total Profit of {top_product['Profit']:,}\n")

# 5, 6, 7 & 8. Create a horizontal bar chart with distinct colors
plt.figure(figsize=(10, 6))

colors = plt.cm.viridis(range(len(plot_data)))

plt.barh(
    plot_data['Product'],
    plot_data['Profit'],
    color=colors,
    edgecolor='black'
)

plt.title('Top 10 Products by Total Profit', fontsize=14)
plt.xlabel('Profit', fontsize=12)
plt.ylabel('Product', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Clean product names to remove any accidental leading/trailing empty spaces
df['Product'] = df['Product'].astype(str).str.strip()

# 1, 2, 3 & 4. Group by Product, sum Profit, sort from highest to lowest, and select the Top 10
top_10_products = (
    df.groupby('Product')['Profit']
    .sum()
    .reset_index()
    .sort_values(by='Profit', ascending=False)
    .head(10)
)

# Print the top-performing product
top_product = top_10_products.iloc[0]
print(f"Highest Profit Product: {top_product['Product']} with Total Profit of {top_product['Profit']:,}\n")

# 5, 6, 7 & 8. Create a horizontal bar chart
plt.figure(figsize=(10, 6))

colors = plt.cm.viridis(range(len(top_10_products)))

plt.barh(
    top_10_products['Product'],
    top_10_products['Profit'],
    color=colors,
    edgecolor='black'
)

# Invert y-axis so the #1 highest profit product appears right at the top
plt.gca().invert_yaxis()

plt.title('Top 10 Products by Total Profit', fontsize=14)
plt.xlabel('Profit', fontsize=12)
plt.ylabel('Product', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Adjust padding to eliminate surrounding empty whitespace around the plot
plt.tight_layout(pad=1.0)
plt.show()
```
